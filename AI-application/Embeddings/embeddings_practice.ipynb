```python
"""In this exercise, you'll create your very first embeddings using the OpenAI API. Normally, to interact with the OpenAI API, you would need an OpenAI API key, and creating embeddings would incur a cost. 

**Instructions**

- Create an OpenAI client.
- Create a request to the Embeddings endpoint, passing the `text-embedding-3-small` model any text you wish.
- Convert the model `response` into a dictionary."""
```

```python
import os
from dotenv import load_dotenv
from openai import OpenAI
```

```python
# Create an OpenAI client
Api_key = os.getenv("OpenAi_API_key")

client =OpenAI(api_key=Api_key)
# Create a request to obtain embeddings
response = client.embeddings.create(
  model="text-embedding-3-small",
  input="This can contain any text."
)

# Convert the response into a dictionary
response_dict = response.model_dump()
print(len(response_dict))
```
