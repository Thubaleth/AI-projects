```python
import os 
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
```

```python
load_dotenv()
api_key = os.getenv("OpenAi_API_key")
```

```python
llm = ChatOpenAI(
    api_key=api_key,
    temperature=0.1,
    max_completion_tokens=100
)
```

```python
response = llm.invoke("what is the capital of france?")
print(response.content)
```
