```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate, FewShotPromptTemplate
```

```python
load_dotenv()
Api_Key = os.getenv("OpenAi_API_key")
```

```python
examples = [
    { 
        "question":"How many courses has jack completed",
        "answer" : "36"

    },
    {
        "question":"How much exprience points does jack have?",
        "answer":"284,320 XP"
    },
    {
        "question":"What technology does jack focus on the most?",
        "answer" : "python"
    }
]
```

```python
#completing the prompt for formatting answers
example_prompt = PromptTemplate.from_template("Question:{question}\n{answer}")

#create the few-shot prompt
prompt_template = FewShotPromptTemplate(
    examples =examples,
    example_prompt=example_prompt,
    suffix = "Question:{input}",
    input_variables = ["input"],

)

prompt = prompt_template.invoke({"input" :"what is jack's favourite technology on Data camp?"})
print(prompt.text)
```

```python
#create an open AI chat llm
llm = ChatOpenAI(
    model = "gpt-4o-mini",
    api_key=Api_Key
)
```

```python
llm_chain = prompt_template | llm
print(llm_chain.invoke({"input": "What is Jack's favorite technology on DataCamp?"}).content)
```
