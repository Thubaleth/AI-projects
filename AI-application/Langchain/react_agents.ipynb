```python
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent
from langchain_community.agent_toolkits.load_tools import load_tools
```

```python
#create llm
llm = ChatOpenAI(
    model = "gpt-4o-mini",
    temperature=0
)
```

```python
#load wikipedia tool
tools = load_tools(["wikipedia"])
```

```python
#create react agent
agent = create_agent(
    llm,
    tools
)
```

```python
#user input
query = "who invented python programmin language?"
```

```python
response = agent.invoke(
    {
        "messages": [
            ("human", query)
        ]
    }
)


```

```python
print(response["messages"][-1].content)
```
