```python
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent
from langchain_community.agent_toolkits.load_tools import load_tools
from langchain_community.tools import DuckDuckGoSearchRun
```

```python
#create llm
llm = ChatOpenAI(
    model = "gpt-4o-mini",
    temperature=0
)
```

```python
#create search tool
search_tool = DuckDuckGoSearchRun()
#add tools
tools = [
    search_tool
]
```

```python
#create agent
agent = create_agent(
    model=llm,
    tools=tools
)
```

```python
query = "Who is current CEO of OpenAI and what are thier recent achievemnets?"
```

```python
response = agent.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": query
            }
        ]
    }
)
```
