```python
import os 
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
```

```python
load_dotenv()
Api_key = os.getenv("OpenAi_API_key")
```

```python
#create model
llm = ChatOpenAI(
    model="gpt-4o-mini",
    api_key=Api_key
)
```

```python
prompt_template = ChatPromptTemplate.from_messages([

    ("system","You are a geography expert that returns the colors present in country flag"),
    ("human","France"),
    ("ai","blue,white,red"),
    ("human","{country}")
])
```

```python
#create chain
llm_chain = prompt_template | llm
#Invoke chain
country = "japan"

response = llm_chain.invoke({
    "country":country
})
```

```python
print(response.content)
```
